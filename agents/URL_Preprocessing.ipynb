```python
import os

BASE_DIR = "/content/drive/MyDrive/Agent"

DATASET_DIR = os.path.join(BASE_DIR, "Dataset")

PREPROCESSED_DIR = os.path.join(BASE_DIR, "Preprocessed_Dataset")

os.makedirs(PREPROCESSED_DIR, exist_ok=True)

print("Dataset folder:", DATASET_DIR)
print("Preprocessed folder:", PREPROCESSED_DIR)
```

```python
print(os.listdir(DATASET_DIR))
```

```python
import pandas as pd
```

```python
legitimate_df = pd.read_csv(
    os.path.join(DATASET_DIR, "Legitimate.csv")
)

malicious_df = pd.read_csv(
    os.path.join(DATASET_DIR, "Malicious.csv")
)

print("Legitimate dataset:", legitimate_df.shape)
print("Malicious dataset:", malicious_df.shape)
```

```python
print("Legitimate columns:", legitimate_df.columns.tolist())
print("Malicious columns:", malicious_df.columns.tolist())

display(legitimate_df.head())
display(malicious_df.head())
```

```python
print("Legitimate missing values:")
print(legitimate_df.isnull().sum())

print("\nMalicious missing values:")
print(malicious_df.isnull().sum())
```

```python
legitimate_df = legitimate_df.dropna(subset=["url"]).copy()
malicious_df = malicious_df.dropna(subset=["url"]).copy()

legitimate_df["url"] = legitimate_df["url"].astype(str).str.strip()
malicious_df["url"] = malicious_df["url"].astype(str).str.strip()

legitimate_df = legitimate_df[legitimate_df["url"] != ""]
malicious_df = malicious_df[malicious_df["url"] != ""]

print("Legitimate after cleaning:", len(legitimate_df))
print("Malicious after cleaning:", len(malicious_df))
```

```python
legitimate_df = legitimate_df.drop_duplicates(subset=["url"])
malicious_df = malicious_df.drop_duplicates(subset=["url"])

print("Legitimate after removing duplicates:", len(legitimate_df))
print("Malicious after removing duplicates:", len(malicious_df))
```

```python
common_urls = set(legitimate_df["url"]) & set(malicious_df["url"])

print("Conflicting URLs:", len(common_urls))

legitimate_df = legitimate_df[
    ~legitimate_df["url"].isin(common_urls)
].copy()

malicious_df = malicious_df[
    ~malicious_df["url"].isin(common_urls)
].copy()

print("Final legitimate URLs:", len(legitimate_df))
print("Final malicious URLs:", len(malicious_df))
```

```python
legitimate_df.to_csv(
    os.path.join(PREPROCESSED_DIR, "Cleaned_Legitimate.csv"),
    index=False
)

malicious_df.to_csv(
    os.path.join(PREPROCESSED_DIR, "Cleaned_Malicious.csv"),
    index=False
)

print("Both cleaned datasets saved successfully!")
```
